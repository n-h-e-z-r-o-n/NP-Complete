# Exploring NP-Complete Problems: Vertex Cover

## 1. Problem Analysis
**Formal Definition:** A vertex cover of an undirected graph $G = (V, E)$ is a subset of vertices $V' \subseteq V$ such that for every edge $(u, v) \in E$, at least one of $u$ or $v$ is in $V'$. The decision problem asks if there exists a vertex cover of size at most $k$.

**Why it is NP-complete:** Vertex Cover is in NP because a proposed set $V'$ can be verified in polynomial time by checking if its size is $\le k$ and if every edge in $E$ is incident to at least one vertex in $V'$. It is NP-hard because the well-known NP-complete problem, Clique, can be reduced to it in polynomial time.

**Real-world Applications:** 
- **Telecommunications:** Placing the minimum number of routing stations (vertices) to monitor all communication links (edges).
- **Cybersecurity:** Selecting the minimum number of nodes to install antivirus software to protect all network connections.


## 2. Reduction Demonstration
**Reducing Vertex Cover to Clique:**
To prove that a problem is NP-complete, we can reduce a known NP-complete problem to it. We will demonstrate the polynomial-time mapping from Vertex Cover to Clique.

**Steps:**
1. Let $G = (V, E)$ be an instance of the Vertex Cover problem, and we want to find if it has a vertex cover of size $k$.
2. Construct the complement graph $\overline{G} = (V, \overline{E})$, where an edge exists in $\overline{G}$ if and only if it does not exist in $G$.
3. $G$ has a vertex cover of size $k$ if and only if $\overline{G}$ has a clique of size $|V| - k$.

**Reasoning:** If $V'$ is a vertex cover in $G$, then no two vertices in $V \setminus V'$ share an edge in $G$ (otherwise the edge wouldn't be covered). Therefore, in the complement graph $\overline{G}$, every pair of vertices in $V \setminus V'$ must share an edge, meaning $V \setminus V'$ forms a clique of size $|V| - k$. This transformation involves creating the complement graph which takes $\mathcal{O}(|V|^2)$ time, proving it is a polynomial-time reduction.

```python
import csv
from collections import defaultdict

def greedy_vertex_cover(edges):
    """
    Greedy approximation algorithm for Vertex Cover.
    Repeatedly picks an edge, adds both endpoints to the cover, 
    and removes all edges incident to either endpoint.
    """
    cover = set()
    # Create a set of edges to modify
    remaining_edges = set(edges)
    
    while remaining_edges:
        # Pick an arbitrary edge
        u, v = remaining_edges.pop()
        # Add both endpoints to the cover
        cover.add(u)
        cover.add(v)
        # Remove all edges incident to u or v
        remaining_edges = { (x, y) for (x, y) in remaining_edges if x != u and x != v and y != u and y != v }
        
    return cover

# Read graph from CSV
edges = []
with open('graph_data.csv', mode='r') as file:
    reader = csv.reader(file)
    next(reader) # Skip header
    for row in reader:
        edges.append((row[0], row[1]))

print(f"Original Edges: {edges}\n")
cover = greedy_vertex_cover(edges)
print(f"Approximate Vertex Cover: {cover}")
```
